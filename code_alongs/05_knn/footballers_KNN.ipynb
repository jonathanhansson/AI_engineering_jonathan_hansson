# KNN footballers

```python
# importing necessary libraries/plugins
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# reading in the CSV
df = pd.read_csv("../../data/footballers.csv")

# What is feature? What is target?
X, y = df[["Mål", "Assist", "Passningsprocent", "Tacklingar", "Luftdueller"]], df["Position"]

# Splitting the dataset into training and test data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)

# Scaling the features. IMPORTANT: Do !NOT! transform X_test. That leaks data.
scaler = MinMaxScaler()
scaled_X_train = scaler.fit_transform(X_train)
scaled_X_test = scaler.transform(X_test)

# Instantiating/Creating an object of the KNN classifier class
football_model_KNN = KNeighborsClassifier(n_neighbors=1)

# We train the model on our training X and training y
football_model_KNN.fit(scaled_X_train, y_train)

# We make a prediction on our test data
y_pred = football_model_KNN.predict(scaled_X_test)

# Here we display the result
y_pred

# This is optional, but gives the players name next to the predicted position
test_players = df.loc[y_test.index, "Spelare"]
for player, prediction in zip(test_players, y_pred):
    print(f"Spelare: {player} - position: {prediction}")

```
